# Intro - NucDB
Decouvrir la base nucleaire embarquee via radioactivedecay.

```python
from rp_toolkit.core.nuc_db import NucDB

db = NucDB()
print('Nombre de nucleides:', db.n_nuclides)
```

```python
cs137 = db.get('Cs-137')
print('T1/2 (ans):', cs137.half_life('y'))
print('Modes:', cs137.decay_modes)
```

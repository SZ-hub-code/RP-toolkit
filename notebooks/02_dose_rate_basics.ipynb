# Dose Rate Basics
Calcul de debit de dose ponctuel et verification de la loi en 1/r^2.

```python
from rp_toolkit.core.dose_rate import DoseScenario, inverse_square_scale

scenario = DoseScenario('Cs-137', activity_bq=2e9, distance_m=1.0)
rate_1m = scenario.dose_rate_uSv_h()
rate_2m = inverse_square_scale(rate_1m, 1.0, 2.0)
print('Rate at 1 m:', rate_1m, 'uSv/h')
print('Rate at 2 m:', rate_2m, 'uSv/h')
```

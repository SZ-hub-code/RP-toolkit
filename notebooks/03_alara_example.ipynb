# ALARA Example
Comparer un scenario de reference et un scenario optimise.

```python
from rp_toolkit.core.alara import AlaraScenario, compare_scenarios

baseline = AlaraScenario(dose_rate_u_sv_h=120.0, time_h=2.0, occupancy_factor=1.0)
optimized = AlaraScenario(dose_rate_u_sv_h=55.0, time_h=1.5, occupancy_factor=0.8)
result = compare_scenarios(baseline, optimized)
result
```

```python
from rp_toolkit.core.alara import allowed_time_for_dose_limit

max_time = allowed_time_for_dose_limit(dose_rate_u_sv_h=30.0, dose_limit_u_sv=100.0)
print('Allowed time (h):', max_time)
```
